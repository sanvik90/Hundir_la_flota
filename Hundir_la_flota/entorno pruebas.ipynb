### Crear tablero

```python
def crear_tablero():
    tablero =np.full((10,10),"🌊")
    return tablero
crear_tablero()
```

### Crear Barco

```python
def crear_barco(eslora):
    proa_fila = np.random.randint(0, 11 - eslora)
    proa_col = np.random.randint(0, 11 - eslora)
    voh = np.random.choice(("a","b")) # un random con dos salidas para elegir si el barco sera horizontal o vertical.
    barco = []
    for i in range(eslora):
        if voh == "a":
            barco.append((proa_fila + i, proa_col))
        else:
            barco.append((proa_fila, proa_col + i))

    return barco
crear_barco(4)
```

### Crear flota

```python
def crear_flota():
    flota = []
    esloras_f = [4, 3, 3, 2, 2, 2]

    for eslora in esloras_f:
        flota.append(crear_barco(eslora))
    for barco in flota:
        for coor in barco:
            if 0 < coor[0]>9 or 0 < coor[1]> 9:
                print("gnerando de nuevo")
                crear_flota()
    coordenadas_vistas = set()
    repetidas = set()

    for barco in flota:
        for coor in barco:
            if coor in coordenadas_vistas:
                repetidas.add(coor)
            coordenadas_vistas.add(coor)

    if repetidas:
        crear_flota()

    
    return flota

crear_flota()
```

### Crear fltoa

```python
def colocar_flota(flota,tablero):
    for barco in flota:
        for pos in barco:
            fila, col = pos
            tablero[fila, col] = "🚢" 
    return tablero
```
